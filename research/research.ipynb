Config Box

```python
example = {
    "key1": "Value1",
    "key2": "Value2"
}
```

```python
example["key1"];
#but example.key1 does not work
```

```python
from box import ConfigBox;
```

```python
example2 = ConfigBox({
    "key1": "Value1",
    "key2": "Value2"
})

example2.key1  #Works
```

Ensurer Annotation

```python
def get_product(x:int, y:int)->int:
    return x*y;
    
```

```python
print(get_product(3, "6")); #gives wrong output
```

```python
from ensure import ensure_annotations
```

```python
@ensure_annotations
def get_prod(x:int, y:int)->int:
    return x*y;
```

```python
get_prod(2, "5"); #throws error
```
